# 🚀 Level 8 — Advanced Pandas

Toss Match WInner New coloumn

```python
def tosswinner(row):
    if row['toss_winner']==row['winner']:
        return 'yes'
    else:
        return 'No'
```

```python
ds['Match-toss Result']=ds[['toss_winner','winner']].apply(tosswinner,axis=1)
ds.columns
```

percentage of Toss winner and match winner

```python
total=ds['Match-toss Result'].size
total
win=ds['Match-toss Result'].value_counts()

yes=int(win.reset_index().iloc[0,1])
no=int(win.reset_index().iloc[1,1])
per=(yes/total)*100
perno=(no/total)*100

print(per.__round__())
print(perno.__round__())
pt.pie([perno,per],labels=['No','Yes'])
pt.title('TOSS WINNER==Match WInner')

```

TossWinner And Match Winner Comparison

```python
toss=ds['toss_winner'].value_counts()
matchwinner=ds['winner'].value_counts()
conc=pd.concat([toss,matchwinner],axis=1)
conc.rename(columns={
    'count':'toss_winner',
    'count':'Match Winner',
   
}, inplace=True)
conc.rename(columns={'Match Winner': 'Toss Winner'}, inplace=True)

print(conc.columns.tolist())
conc.columns = ['Toss Winner', 'Match Winner']
conc

```
